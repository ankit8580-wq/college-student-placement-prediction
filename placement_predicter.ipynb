```python
df = pd.read_csv("/content/college_student_placement/college_student_placement_dataset.csv")
```

```python
df.head()
```

```python
df = df.iloc[:,1:]     #(Preprocess) removed college_id and retrieve it again we just have to read the dataset again
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
df['Placement'] = df['Placement'].map({'Yes': 1, 'No': 0})   #convert 'Placement' values from string to number i.e., Yes -> 1 and No -> 0
```

```python
plt.scatter(df['CGPA'],df['IQ'],c=df['Placement'])   # (EDA)  it only takes two features, since it is plooting in x and y axis
```

```python
#c=df['Placement' -> this is coloring those students who got the placement.
#Here, yellow is showing placement(1)
# purple is showing not placed(0).
```

```python
df = df.drop('Prev_Sem_Result', axis=1)   #Remove Prev_Sem_Result
```

```python
df.head()
```

```python
#Step1 : Extract input and output columns
x = df.iloc[:,0:2]  # IQ and CGPA
y = df.iloc[:,-1]   #Placement
```

```python
x
```

```python
y
```

```python
y.shape
```

```python
x.shape
```

```python
#Step3 : Train test splits
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size = 0.1)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
#Scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
#Step4: Train the model
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression()   #we made 'clf' as object
```

```python
clf.fit(X_train,y_train)     #model training
```

```python
#Step 5: Evaluate the model
y_pred = clf.predict(X_test)
```

```python
y_test
```

```python
#To check the accuracy between our model and actual output
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X_train, y_train.values, clf=clf, legend=2)
```
